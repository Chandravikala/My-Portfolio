# Load Covid19 Data from API (https://api.covidtracking.com) into PostgreSQL Database

```python
# download necessary libraries 
# requests - to process json responses
# pandas - to convert json responses into dataframe
# sqlalchemy - to establish the connection in postgresql and create the table load
```

```python
import requests
import pandas
import sqlalchemy
```

```python
# pass on the api url that contains data in json format
url = 'https://api.covidtracking.com/v2/states/ca/daily/simple.json'
response = requests.get(url)
# printing the response status code
# status code 200 - 
print(response.status_code)
```

```python
# printing the data as in json format
responseData = response.json()
print(responseData)
```

```python
# converting json repsonses into dataframe using json_normalize method
df = pandas.json_normalize(responseData, 'data')
print(df)
```

```python
# viewing first 10 rows of the created dataframe
df.head(10)
```

```python
df.info()
df.describe()
```

```python
# Define your database connection details
# create a DB name COVID19 in postgresql and then run this code.
DATABASE_URL = "postgresql+psycopg2://username:password@server:port/COVID19_new"

# Create an SQLAlchemy engine
engine = sqlalchemy.create_engine(DATABASE_URL)
with engine.begin() as connection:
    df.to_sql(name='COVID_19_CA', con=connection, index=False, if_exists='append')
```
